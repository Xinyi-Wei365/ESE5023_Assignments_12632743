```python
import random

# 做 M1：5 行 10 列
M1 = []
for i in range(5):
    row = []
    for j in range(10):
        num = random.randint(0, 50)
        row.append(num)
    M1.append(row)

# 做 M2：10 行 5 列
M2 = []
for i in range(10):
    row = []
    for j in range(5):
        num = random.randint(0, 50)
        row.append(num)
    M2.append(row)

# 2.2 矩阵乘法
def Matrix_multip(M1, M2):
    result = []                     # 结果矩阵：5 行 5 列
    for i in range(5):              # 结果的每一行
        row = []
        for j in range(5):          # 结果的每一列
            s = 0                   # 累加器，先清零
            for k in range(10):     # M1 的列数 = M2 的行数 = 10
                s = s + M1[i][k] * M2[k][j]   # 对应相乘、累加
            row.append(s)
        result.append(row)
    return result

M3 = Matrix_multip(M1, M2)
print(M3)
```
