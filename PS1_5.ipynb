```python
import matplotlib.pyplot as plt
def count_solutions(x):
    count = 0 
    for op1 in ["+","-",""]:
        for op2 in ["+","-",""]:
            for op3 in ["+","-",""]:
                for op4 in ["+","-",""]:
                    for op5 in ["+","-",""]:
                        for op6 in ["+","-",""]:
                            for op7 in ["+","-",""]:
                                for op8 in ["+","-",""]:
                                    expr = "1"+ op1 +"2"+ op2 +"3" + op3 +"4"+ op4 +"5"+ op5+"6" +op6+"7"+op7+"8"+op8+"9"
                                    if eval(expr) == x:
                                        count = count + 1
 
    return count

Total_solutions = []
for i in range(1, 101):
    Total_solutions.append(count_solutions(i))

plt.figure(figsize=(10, 5))
plt.plot(range(1, 101), Total_solutions, marker="o", markersize=3)
plt.xlabel("Integer i (1 to 100)")
plt.ylabel("Total number of solutions")
plt.title("Total_solutions for expressions made of 123456789")
plt.grid(True, alpha=0.3)
plt.show()

maximum = max(Total_solutions)
minimum = min(Total_solutions)
print("Maximum =", maximum, "at i =", [i + 1 for i in range(100) if Total_solutions[i] == maximum])
print("Minimum =", minimum, "at i =", [i + 1 for i in range(100) if Total_solutions[i] == minimum])
```
